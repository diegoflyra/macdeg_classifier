# Análise de um lote

Roda sozinho depois do job array (ou à mão, apontando `LOTE`). Junta as 20 tarefas de cada configuração e calcula:

- **F2 médio por base** (métrica principal) com IC95 por paciente, limiar sempre escolhido em outras dobras;
- AUROC, AUPRC, sensibilidade, **especificidade em normais e em outras doenças** (e por doença);
- **réguas**: só a base de origem e só a idade (o modelo precisa superá-las);
- **veredito pareado** contra a referência: *ganhou*, *empatou* ou *perdeu*. Só "ganhou" permite dizer "melhorou";
- **resultado aninhado** (quando há mais de uma configuração): a estimativa honesta de "rodar a grade e escolher a melhor";
- **distribuição das notas** de normais, outras doenças, drusas sem diagnóstico e DMRI.

Tudo vai para `resultados/<lote>/` na branch `resultados` do GitHub (só IDs nossos, nenhuma imagem) e um resumo vai ao W&B.

```python
LOTE = ""        # pasta do lote
DADOS = ""       # conjunto (para as réguas de origem e idade)
ENVIAR = True    # False = só calcula, sem GitHub
```

```python
import json
from pathlib import Path
import pandas as pd
from macdeg.ambiente import load_dotenv
from macdeg.analysis import analyze, ranking

load_dotenv()   # $MACDEG_ENV: o lançador aponta para o .env do clone no $HOME (só o caminho)
pd.set_option("display.max_columns", None, "display.width", 220)
resultado = analyze(Path(LOTE), DADOS or None, push=ENVIAR)
```

```python
if resultado.get("configuracoes"):
    display(ranking(resultado).round(4))
```
